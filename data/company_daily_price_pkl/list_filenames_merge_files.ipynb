```python
import os

folder = "E:/factors_affecting_stock_price/data/company_daily_price_pkl"
keyword = "naver"

files = [f for f in os.listdir(folder) if keyword in f]

# print(files)
```

```python
import pandas as pd
import re

# File list provided by user context
# file_names = [
#     "hyunmotor_10min_price_20240125.csv",
#     "005380_hyunmotor_10min_price_20220620.csv",
#     "005380_hyunmotor_10min_price_20220622.csv",
#     "hyunmotor_10min_price_20231011.csv"
# ]
file_names = files[155:]

# Function to extract date from filename for sorting
def extract_date(filename):
    # Search for an 8-digit sequence (YYYYMMDD)
    match = re.search(r'\d{8}', filename)
    return match.group(0) if match else "00000000"

# Sort the file list by date
sorted_files = sorted(file_names, key=extract_date)
print("Sorted File List:")
print(sorted_files)

dfs = []

for filename in sorted_files:
    # Read the file
    try:
        df = pd.read_csv(filename)
    except:
        print("***************************************************************************")
        print(filename)
        print("***************************************************************************")
    
    # Check for unnamed index column. 
    # Based on snippets, some files have an index column as the first column (often Unnamed: 0).
    # We check if 'date' is in columns. If 'date' is not the first column, we might need to adjust.
    if "date" in df.columns:
        # If 'date' is a column, check if there's a junk column before it (like Unnamed: 0)
        if df.columns[0].startswith('Unnamed'):
            df = df.drop(df.columns[0], axis=1)
    
    dfs.append(df)

# Concatenate all dataframes
merged_df = pd.concat(dfs, ignore_index=True)

# Ensure data is sorted chronologically by date and time
# First, convert to datetime to ensure proper sorting
merged_df['datetime'] = pd.to_datetime(merged_df['date'] + ' ' + merged_df['time'])
merged_df = merged_df.sort_values(by='datetime')

# Drop the helper datetime column
merged_df = merged_df.drop(columns=['datetime'])

# Save the merged file
output_filename = "hyunmotor_10min_price_merged5.csv"
merged_df.to_csv(output_filename, index=False)

print(f"\nMerged DataFrame Info:")
print(merged_df.info())
print(f"\nHead of merged data:")
print(merged_df.head())
```
